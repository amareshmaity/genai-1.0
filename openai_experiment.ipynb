```python
client = OpenAI()

completion = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {
            "role": "user",
            "content": "Write a one-sentence bedtime story about a unicorn.",
        }
    ],
)

print(completion)
```

```python
completion.choices[0].message
```

```python
completion.choices[0].message.role
```

```python
completion.choices[0].message.content
```

## Chat conversation

```python
## Chat conversation

messages=[
    {
        "role": "user",
        "content": "Write a one-sentence bedtime story about a unicorn.",
    },
    {
        "role": "assistant",
        "content": "Under a shimmering moon, a gentle unicorn whispered dreams of stardust to sleepy flowers, lulling the night into a peaceful lullaby.",        
    }
]
```

```python
messages=[
    {
        "role": "user",
        "content": "Hi, my name is Amaresh, I am an AI engineer.",
    }
]

response = client.responses.create(
    model="gpt-4.1-mini",
    input=messages
)

response.output_text
```

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    input="who am I?"
)

response.output_text
```

```python
messages=[
    {
        "role": "user",
        "content": "Hi, my name is Amaresh, I am an AI engineer.",
    },
    {
        "role": "assistant",
        "content": "Hello Amaresh! It's great to meet you. How can I assist you today?",
    },

]

messages.append({
            "role": "user",
            "content": "who am I?"
        })

response = client.responses.create(
    model="gpt-4.1-mini",
    input=messages
)

response.output_text
```
